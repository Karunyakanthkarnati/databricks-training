## filter() or where()
Both `filter()` and `where()` are used to subset rows in a DataFrame based on a condition. They are interchangeable in Spark and Databricks.

**Example:**
python
df.filter(df.age > 21)
df.where(df.age > 21)

Both lines return rows where the `age` column is greater than 21.

```python
data = [(1,'raj','Male',2000),(2,'veky','Male',3000),(3,'srinidhi','Female',4000)]
schema = ['id','name','gender','salary']
df = spark.createDataFrame(data, schema)
df.show()
```

```python
from pyspark.sql.functions import col
df.filter(col('salary') > 2000).show()
```

```python
df.where((col("gender") == 'Male') & (col("salary") > 2000)).show()
df.where((col("gender") == 'Male') | (col("salary") > 2000)).show()
```

## distinct()
The `distinct()` function returns a new DataFrame with duplicate rows removed.

**Example:**
python
df.distinct().show()

This displays only unique rows from the DataFrame.

```python
data = [(1,'raj','Male',2000),(2,'veky','Male',3000),(2,'veky','Male',3000),(3,'srinidhi','Female',4000)]
schema = ['id','name','gender','salary']
df2 = spark.createDataFrame(data, schema)
df2.show()
```

```python
df2.distinct().show()
```

## dropDuplicates()
The `dropDuplicates()` function removes duplicate rows from a DataFrame, similar to `distinct()`, but it also allows you to specify columns to consider for identifying duplicates.

**Example:**
python
#### Removes duplicate rows based on all columns
df2.dropDuplicates().show()

#### Removes duplicate rows based only on the 'name' and 'gender' columns
df2.dropDuplicates(['name', 'gender']).show()

```python
df2.dropDuplicates(["gender"]).show()
df2.dropDuplicates(["gender","name"]).show()
```

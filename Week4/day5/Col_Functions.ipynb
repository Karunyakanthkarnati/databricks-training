## when() & otherwise()
The `when()` function in PySpark is used for conditional expressions, similar to SQL's `CASE WHEN`. It allows you to specify a condition and the value to return if the condition is true. The `otherwise()` function specifies the value to return if none of the `when()` conditions are met.

**Example:**

python
from pyspark.sql.functions import when, col

df.withColumn(
    "category",
    when(col("score") >= 90, "Excellent")
    .when(col("score") >= 75, "Good")
    .otherwise("Average")
)

```python
data = [(1,'raj','M',2000),(2,'zil','',3000),(3,'srini','F',4000)]
schema = ['id','name','gender','salary']
df = spark.createDataFrame(data, schema)
df.show()
```

```python
from pyspark.sql.functions import *
df1 = df.select("id","name",\
                when(col('gender')=='M','Male')\
                .when(df.gender == 'F','Female')\
                .otherwise('unknown').alias("gender")\
                )
df1.show()
```

## Alias()
The `alias()` function in PySpark is used to rename a column or assign an alias to an expression in a DataFrame. This is useful for making column names more readable or for referencing columns in subsequent operations.

**Example:**

python
from pyspark.sql.functions import col

df.select(col("name").alias("employee_name")).show()

```python
df.select(col("id").alias('emp_id')).show()
```

## asc() & desc()
The `asc()` and `desc()` functions in PySpark are used to sort DataFrame columns in ascending or descending order, respectively.

**Example:**

python
from pyspark.sql.functions import asc, desc

### Sort by salary in ascending order
df.orderBy(asc("salary")).show()

### Sort by salary in descending order
df.orderBy(desc("salary")).show()

```python
df.sort(col("name").desc()).show()
```

## cast()
The `cast()` function in PySpark is used to convert the data type of a column to another data type.

**Example:**

python
from pyspark.sql.functions import col

df.select(col("salary").cast("string")).show()

```python
df.select(col("name"),col("salary").cast('string')).printSchema()
```

## like()
The `like()` function in PySpark is used to filter rows based on pattern matching, similar to SQL's `LIKE` operator.

**Example:**

python
from pyspark.sql.functions import col

### Select rows where the name starts with 'r'
df.filter(col("name").like("r%")).show()

```python
df.filter(col("name").like('s%')).show()
```

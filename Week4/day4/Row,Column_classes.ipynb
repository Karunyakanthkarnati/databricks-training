## Row() class
The `Row` class in PySpark is used to represent a record in a DataFrame. Each field in the Row can be accessed like an attribute.

**Example:**
python
from pyspark.sql import Row

row = Row(name="Alice", age=30)
print(row.name)  # Output: Alice
print(row['age'])  # Output: 30


```python
from pyspark.sql import Row

row = Row('kanth',19)
print(row)
print(row[0])
print(row[0] + " " + str(row[1]))
      
```

```python
row1 = Row(name = 'kanth',age = 19)
print(row1.name)
print(row1.age)
```

```python
a = Row(name = 'Karunya',age = 19)
b = Row(name = 'Kanth',age = 23)
data = [a,b]
df = spark.createDataFrame(data)
df.show()
```

```python
data = [Row(name = 'Karunya', prop = Row(age = 19, height = 5.10)),Row(name = 'Kanth', prop = Row(age = 23, height = 5.8))]
df = spark.createDataFrame(data)
df.show()
df.printSchema()
```

```python
person = Row('name', 'age')
p1 = person('Kanth', 23)
p2 = person('Karunya', 19)
print(p1.name)
print(p2.age)
```

## column() class
The `Column` class in PySpark represents a column expression in a DataFrame. It is used to perform operations on DataFrame columns, such as arithmetic, logical operations, and transformations.

**Example:**
python
from pyspark.sql.functions import col

df.select(col("age") + 1).show()

```python
from pyspark.sql.functions import lit , col
data = [('kanth',19,3000000),('karunya',20,400000)]
schema = ['name','age','salary']
df = spark.createDataFrame(data,schema)
df1 = df.withColumn('nationality',lit('India'))
df1.show()
df1.printSchema()
```

```python
df1.select(df["age"]).show()
df1.select(col("name")).show()
df1.select(df.salary).show()
```

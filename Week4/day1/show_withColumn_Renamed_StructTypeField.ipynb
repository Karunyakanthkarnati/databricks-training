## withcolumn() method

```python
from pyspark.sql.functions import *  #imported col and lit 

data = [(1,'ravi',5000),(2,'kumar',6000),(3,'ram',7000),(4,'shyam',8000)]
schema = ['id','name','salary']
df = spark.createDataFrame(data, schema) # created dataframe
df.show()
```

```python
df1 = df.withColumn('salary',col('salary')*10) #changed the salary column
df1.show()
```

```python
df2 = df.withColumn('bonus',lit(10000)) #added bonus column
df2.show()
df3 = df2.withColumnRenamed('salary','new_salary') #renamed salary column
df3.show()
```

```python
df4 = df.withColumn('salary',col("salary").cast('int')) #changed datatype of salary column
df4.show()
df4.printSchema()
```

## withColumnRenamed()

```python
df_rename = df.withColumnRenamed("salary","salary_amount") #renamed salary column
df_rename.show()
```

## StructType() & StructField

```python
from pyspark.sql.types import * # imported StructType and StructField
data = [(1,'ravi',5000),(2,'kumar',6000),(3,'ram',7000),(4,'shyam',8000)]
schema = StructType([StructField('id',IntegerType(),True),\
            StructField('name',StringType(),True),\
            StructField('salary',IntegerType(),True)]) # schema defined using StructType
df = spark.createDataFrame(data,schema)
df.display()
df.printSchema()
```

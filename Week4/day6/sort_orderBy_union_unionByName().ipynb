## Sort() & orderby()
- **sort()**: A method used to sort data in ascending or descending order. In Python, `list.sort()` sorts a list in place, while `sorted()` returns a new sorted list.
- **orderby()**: Commonly used in SQL or PySpark to specify the order of rows in query results. Example: `SELECT * FROM table ORDER BY column ASC;` or `df.orderBy("column")` in PySpark.

```python
data = [(1,'raj','Male',2000,'IT'),(2,'veky','Male',3000,'HR'),(3,'srinidhi','Female',5000,'sales'),(4,'surya','Male',5000,'IT'),(5,'sita','Female',6000,'HR')]
schema = ['id','name','gender','salary','dept']
df = spark.createDataFrame(data, schema)
df.sort(df.dept,df.salary.desc()).show()
df.orderBy(df.dept.desc(),df.salary).show()
```

## Union() & UnionAll()
- **union()**: Combines the rows of two DataFrames with the same schema, including duplicate rows. In PySpark: `df1.union(df2)`.
- **unionAll()**: Deprecated in recent versions of PySpark; use `union()` instead. Previously, it was an alias for `union()` and combined all rows, including duplicates.

```python
data1 = [(1,'raj','Male',2000,'IT'),(2,'veky','Male',3000,'HR'),(3,'srinidhi','Female',5000,'sales')]
schema1 = ['id','name','gender','salary','dept']
df1 = spark.createDataFrame(data1, schema1)

data2 = [(2,'veky','Male',3000,'HR'),(4,'surya','Male',5000,'IT'),(5,'sita','Female',6000,'HR')]
schema2 = ['id','name','gender','salary','dept']
df2 = spark.createDataFrame(data2, schema2)
df1.union(df2).show()
df1.union(df2).distinct().show()
```

## unionByName()
- **unionByName()**: Combines two DataFrames by matching column names, not positions. Useful when DataFrames have columns in different orders or when some columns are missing. Example: `df1.unionByName(df2)`.

```python
data1 = [(1,'raj','Male'),(2,'veky','Male'),(3,'srinidhi','Female')]
schema1 = ['id','name','gender']
df1 = spark.createDataFrame(data1, schema1)

data2 = [(2,'venky',3000,'HR'),(4,'surya',5000,'IT'),(5,'sita',6000,'HR')]
schema2 = ['id','name','salary','dept']
df2 = spark.createDataFrame(data2, schema2)
df1.unionByName(df2, allowMissingColumns =True).show()
```

## groupBy()
The `groupBy()` function in PySpark is used to group rows in a DataFrame based on the values of one or more columns. After grouping, you can apply aggregate functions such as `count()`, `sum()`, `avg()`, etc., to compute summary statistics for each group.

**Example:**

python
#### Group by 'department' and count the number of employees in each department
df.groupBy("department").count().show()

```python
data = [(1,'raj','Male',2000,'IT'),(2,'veky','Male',3000,'HR'),(3,'srinidhi','Female',5000,'sales'),(4,'surya','Male',5000,'IT'),(5,'sita','Female',6000,'HR'),(6,'ram','Male',7000,'sales'),(7,'laxmi','Female',8000,'IT'),(8,'lakshmi','Female',9000,'HR'),(9,'sravani','Female',10000,'sales')]
schema = ['id','name','gender','salary','dept']
df = spark.createDataFrame(data, schema)
df.show()
df.groupBy("dept").count().show()
df.groupBy("dept").max("salary").show()
df.groupBy("dept").min("salary").show()
df.groupBy("dept").avg("salary").show()
df.groupBy("dept").sum("salary").show()
```

## agg()
The `agg()` function in PySpark allows you to perform multiple aggregate operations on grouped data at once. You can specify different aggregate functions for different columns within the same call.

**Example:**

python
from pyspark.sql import functions as F

#### Group by 'dept' and calculate the average and maximum salary in each department
df.groupBy("dept").agg(
    F.avg("salary").alias("avg_salary"),
    F.max("salary").alias("max_salary")
).show()

```python
from pyspark.sql.functions import *
df.groupBy("dept").agg(count('id').alias('count'),\
    max('salary').alias('max_salary'),\
        min('salary').alias('min_salary'),\
            avg('salary').alias('avg_salary'),\
                sum('salary').alias('sum_salary')\
                    ).show()
```

## select()
The `select()` function in PySpark is used to select one or more columns from a DataFrame. You can also use expressions to create new columns or modify existing ones.

**Example:**

python
#### Select 'name' and 'salary' columns
df.select("name", "salary").show()

#### Select 'name' and create a new column 'salary_in_k' by dividing salary by 1000
from pyspark.sql.functions import col
df.select("name", (col("salary")/1000).alias("salary_in_k")).show()

```python
df.select("name",df["dept"],df.gender,col("salary")).show() # all representations are the same
df.select(["id",'name']).show() # can give in list
```

```python
df.select('*').show() # select all columns
```

```python
df.select([col for col in df.columns]).show() # columns are selected by iterating through the list
```

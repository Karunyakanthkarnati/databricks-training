## ArrayType()

```python
from pyspark.sql.types import *

data = [('abc',[1,2,3]),('def',[4,5,6]),('ghi',[7,8,9])]
schema = StructType([StructField('id',StringType()),StructField('value',ArrayType(IntegerType()))])
df = spark.createDataFrame(data,schema)
df.show()
df.printSchema()
```

fetch value from an array as a new column

```python
from pyspark.sql.functions import *
df1 = df.withColumn('First value',col('value')[0])
df1.show()
df1.printSchema()
```

## array()
combine columns to array

```python
data = [(1,2),(3,4)]
schema = ['num1','num2']
df2 = spark.createDataFrame(data,schema)
df2 = df2.withColumn('numbers',array(col("num1"),col("num2")))
df2.show()
```

## explode()
creates a new row for each element in an array column

```python
data = [(1,'kanth',['python','streemlit']),(2,'raju',['sql','pyspark'])]
schema = ['id','name','skills']
df = spark.createDataFrame(data,schema)
df.show()
df.printSchema()
```

```python
from pyspark.sql.functions import *
df1 = df.withColumn('skill',explode(col('skills')))
df1.show()
df1.printSchema()
```

## split()
splitting a string column by delimiter and returns array type

```python
data = [(1,'kanth','python,streemlit'),(2,'raju','sql,pyspark')]
schema = ['id','name','skills']
df = spark.createDataFrame(data,schema)
df.show()
df.printSchema()
```

```python
df1 = df.withColumn('sKillsArray',split(col('skills'),(',')))
df1.show()
df1.printSchema()
```

## array_contains()
checks if an array column contains a value.
null if array is null
true if array contains value
false if it does not

```python
data = [(1,'kanth',['python','streemlit']),(2,'raju',['sql','pyspark'])]
schema = ['id','name','skills']
df = spark.createDataFrame(data,schema)
df.show()
df.printSchema()
```

```python
from pyspark.sql.functions import *
df1 = df.withColumn('sql',array_contains(col("skills"),'sql'))
df1.show()
df1.printSchema()
```

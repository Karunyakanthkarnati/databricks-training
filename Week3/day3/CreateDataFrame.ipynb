## creating schema datatypes manually

```python
from pyspark.sql.types import *

data = [{'id':1,'name':'Raj'},{'id':2,'name':'Ravi'}]
schema = StructType([StructField('id',dataType = IntegerType()),StructField('name',dataType = StringType())])
df = spark.createDataFrame(data,schema)
df.display()
df.printSchema()
```

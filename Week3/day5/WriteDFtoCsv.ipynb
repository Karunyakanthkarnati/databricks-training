## Reading or create DataFrame before writing

```python
data = [(1, "John"), (2, "Mary"), (3, "Mike"), (4, "Jane")]
schema = ("id","name")

df = spark.createDataFrame(data, schema)
df.display()
```

### method 1

```python
df.write.csv(path = "/Volumes/workspace/default/databricks2027/output",header = True,mode = "overwrite")
```

### method 2

```python
df.write.format("csv").option('header',True).mode("overwrite").save("/Volumes/workspace/default/databricks2027/output")
```

## to check the content we can use read option

```python
df2 = spark.read.option('header',True).csv("/Volumes/workspace/default/databricks2027/output")
df2.display()
```

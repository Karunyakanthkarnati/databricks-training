## Reading the file

```python
df = spark.read.format("csv").option('header',True).option('inferSchema',True).load("/Volumes/workspace/default/databricks2027/empData.csv")
df.display()
```

```python
df.printSchema()
```

```python
my_schema = '''
  emp_id String,
  name string,
  salary integer,
  address string'''
```

```python
df1 = spark.read.format('csv').option('header','true').schema(my_schema).load('/Volumes/workspace/default/databricks2027/empData.csv')
df1.display()
```

```python
df_csv = spark.read.format("csv").option('header','true').option('inferSchema','true').load('/Volumes/workspace/default/databricks2027/Big Sales.csv')
df_csv.display()
```

## Display only 3 columns using select in pyspark

```python
df_csv.select('Item_Identifier','Item_Fat_Content','Item_Type').display()
```

```python
from pyspark.sql.functions import *

df_csv.select(col("Item_Identifier"),col("Item_Fat_Content"),col("Item_Type")).display()
```

## Using Alias

```python
df_csv.select(col("Item_Identifier").alias("Item_Id")).display()
```

### sorting the values

```python
df_csv.sort(col("Item_Weight").desc()).display()
```

```python
df_csv.sort(["Item_Visibility","Item_MRP"],ascending=[0,0]).display()
```

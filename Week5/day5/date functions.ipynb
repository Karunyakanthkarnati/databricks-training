
## Date Functions

Date functions are used to manipulate and extract information from date and timestamp values. Common examples include:

- `current_date()` — Returns the current date.
- `current_timestamp()` — Returns the current timestamp.
- `date_add(start_date, days)` — Adds a specified number of days to a date.
- `date_sub(start_date, days)` — Subtracts a specified number of days from a date.
- `datediff(end_date, start_date)` — Returns the difference in days between two dates.
- `year(date)` — Extracts the year from a date.
- `month(date)` — Extracts the month from a date.
- `day(date)` — Extracts the day of the month from a date.
- `hour(timestamp)` — Extracts the hour from a timestamp.
- `minute(timestamp)` — Extracts the minute from a timestamp.
- `second(timestamp)` — Extracts the second from a timestamp.

Example:
sql
SELECT current_date(), year(current_date()), month(current_date()), day(current_date());


```python
from pyspark.sql.functions import *
df = spark.range(3)
df1 = df.withColumn("currdate",current_date())
df1.show()
```

- `date_format(date, format)` — Formats a date or timestamp according to the specified pattern. For example, `date_format(current_date(), 'yyyy-MM-dd')` returns the date as a string in "YYYY-MM-DD" format.

```python
df2 = df1.withColumn('dateformat',date_format(df1.currdate,'dd.mm.yyyy'))
df2.show()
df2.printSchema()
```

- `to_date(date_string, format)` — Converts a string to a date using the specified format. For example, `to_date('2026-07-23', 'yyyy-MM-dd')` returns a date value.

```python
df3 = df2.withColumn("date",to_date(df2.currdate,'dd.mm.yyyy'))
df3.show()
df3.printSchema()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from datetime import datetime
df = spark.createDataFrame([('2019-07-01','2019-09-14')],['start','end'])
df.show()
```

```python
df1 = df.withColumn("datediff",datediff(df.end,df.start))
df1.show()
```

```python
df.withColumn('monthsb/w',months_between(df.end,df.start)).show()
df.withColumn("addmonths",add_months(df.start,4)).show()
df.withColumn("dateadd",date_add(df.end,15)).show()
df.withColumn("year",year(df.start)).show()
df.withColumn("month",month(df.start)).show()
df.withColumn("day",dayofmonth(df.start)).show()
df.withColumn("dayofweek",dayofweek(df.start)).show()
df.withColumn("dayofyear",dayofyear(df.start)).show()
df.withColumn("weekofyear",weekofyear(df.start)).show()
```

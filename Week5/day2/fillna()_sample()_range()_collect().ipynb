## fill() & fillna()
`fill()` and `fillna()` are used in PySpark DataFrames to replace null or missing values.

- `fillna(value, subset=None)`: Replaces null values with the specified value. You can specify columns using `subset`.
- `fill(value, subset=None)`: Alias for `fillna()`.

**Example:**
python
df.fillna(0)  # Replace all nulls with 0
df.fillna({'age': 25, 'name': 'unknown'})  # Replace nulls in 'age' and 'name' columns
df.fillna(0, subset=['age'])  # Replace nulls in 'age' column only



```python
data = [(1,"ravi",None,"IT"),(2,"raj",5000,None),(3,"ram",6000,"HR"),(4,None,7000,"IT"),(5,"raj",8000,None),(6,"ram",9000,"HR")]
schema = ["id", "name", "salary", "dept"]
df = spark.createDataFrame(data,schema)
df.show()
```

```python
df.fillna(0).show()  ## filling null valves with 0 for int columns
df.fillna("unknown").show()  ## filling null valves with "unknown" for string columns
df.fillna(0,subset=["salary"]).show()  ## filling null valves with 0 for salary column
df.fillna("unknown",["dept"]).show()
df.na.fill(0).show()
```

## sample()
`sample()` in PySpark is used to randomly sample rows from a DataFrame.

- `df.sample(withReplacement, fraction, seed=None)`: 
  - `withReplacement`: Whether sampling is with replacement (True/False).
  - `fraction`: Fraction of rows to return (e.g., 0.1 for 10%).
  - `seed`: Optional random seed for reproducibility.

**Example:**
python
df.sample(False, 0.5).show()  # Randomly sample 50% of the rows without replacement
df.sample(True, 0.2, seed=42).show()  # Sample 20% of the rows with replacement, using seed 42

## Range()
`range()` in PySpark creates a DataFrame with a single column of consecutive integers.

- `spark.range(start, end, step, numPartitions=None)`:
  - `start`: Starting value (default 0).
  - `end`: Ending value (exclusive).
  - `step`: Step size (default 1).
  - `numPartitions`: Number of partitions (optional).

**Example:**
python
spark.range(5).show()  # Creates a DataFrame with values 0 to 4
spark.range(2, 10, 2).show()  # Values: 2, 4, 6, 8

```python
df1 = spark.range(start=1,end=101)
df1.show()
```

```python
df1.sample(fraction = 0.1).show()
df1.sample(fraction=0.1).show()
```

```python
df2 = df1.sample(fraction=0.1,seed=1002)
df3 = df1.sample(fraction=0.1,seed=1002)
display(df2)
display(df3)
```

## collect()
`collect()` in PySpark retrieves all rows from a DataFrame as a list of Row objects.

- `df.collect()`: Returns all rows in the DataFrame to the driver.

**Example:**
python
rows = df.collect()  # Collects all rows as a list
for row in rows:
    print(row)

```python
data = [(1,"ravi",5000),(2,"raj",6000),(3,"ram",7000),(4,"raj",8000),(5,"ram",9000)]
schema = ["id", "name", "salary"]
df = spark.createDataFrame(data,schema)
df.show()
```

```python
listrows = df.collect()
print(listrows)
print(listrows[0])
print(listrows[1][2])
```

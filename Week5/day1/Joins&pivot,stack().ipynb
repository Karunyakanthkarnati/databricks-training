## Joins()
Joins are used to combine rows from two or more DataFrames based on a related column between them. Common types of joins in PySpark include:

- **Inner Join**: Returns records with matching values in both DataFrames.
- **Left Join (Left Outer Join)**: Returns all records from the left DataFrame, and matched records from the right DataFrame.
- **Right Join (Right Outer Join)**: Returns all records from the right DataFrame, and matched records from the left DataFrame.
- **Full Join (Full Outer Join)**: Returns all records when there is a match in either left or right DataFrame.
- **Cross Join**: Returns the Cartesian product of both DataFrames.
- **Left Anti Join**: Returns only records from the left DataFrame that do not have a match in the right DataFrame.
- **Left Semi Join**: Returns only records from the left DataFrame that have a match in the right DataFrame.

Example PySpark syntax:
python
df1.join(df2, df1.key == df2.key, 'inner')
df1.join(df2, df1.key == df2.key, 'left')
df1.join(df2, df1.key == df2.key, 'right')
df1.join(df2, df1.key == df2.key, 'full')
df1.crossJoin(df2)
df1.join(df2, df1.key == df2.key, 'left_anti')
df1.join(df2, df1.key == df2.key, 'left_semi')

```python
data1 = [(1,'raj',5000,2),(2,'rajesh',6000,4),(3,'prakash',7000,5),(4,'venu',8000,1),(5,'ramu',9000,3)]
schema1 = ["id","name","salary","dept"]
data2 = [(1,'IT'),(2,'HR'),(3,'FINANCE'),(4,'MARKETING'),(6,'SALES')]
schema2 = ["dept","dept_name"]
empdf = spark.createDataFrame(data1,schema1)
deptdf = spark.createDataFrame(data2,schema2)
empdf.show()
deptdf.show()
```

```python
empdf.join(deptdf, empdf.dept == deptdf.dept,'inner').show()
empdf.join(deptdf, empdf.dept == deptdf.dept,'left').show()
empdf.join(deptdf,empdf.dept == deptdf.dept,'right').show()
empdf.join(deptdf, empdf.dept == deptdf.dept,'full').show()
```

```python
empdf.join(deptdf, empdf.dept == deptdf.dept,'leftsemi').show()
empdf.join(deptdf,empdf.dept == deptdf.dept,'leftanti').show()
```

### Self Join

A self join is a join in which a DataFrame is joined with itself. This is useful for comparing rows within the same DataFrame, such as finding employees who share the same department.

Example PySpark syntax:
python
empdf.alias("e1").join(empdf.alias("e2"), col("e1.dept") == col("e2.dept"), "inner").show()

## pivot()
The `pivot()` function in PySpark is used to rotate DataFrame data, turning unique values from one column into multiple columns in the output. This is useful for creating summary tables or cross-tabulations.

**Syntax:**
python
df.groupBy(grouping_column).pivot(pivot_column).agg(aggregation_function)


**Example:**
Suppose you want to see the total salary by department for each employee name:
python
empdf.groupBy("name").pivot("dept").sum("salary").show()


```python
data = [(1,'raj','male','IT'),(2,'rajesh','male','HR'),(3,'lakshmi','female','FINANCE'),(4,'venu','male','MARKETING'),(5,'latha','female','SALES'),(6,'ravi','male','IT'),(7,'rakesh','male','HR'),(8,'prasanthi','female','FINANCE'),(9,'sandhya','female','MARKETING'),(10,'ram','male','SALES')]
schema = ["id","name","gender","dept"]
df = spark.createDataFrame(data,schema)
df.show()
```

```python
df.groupby("dept").pivot("gender").count().show()
df.groupby('gender').pivot("dept").count().show()
df.groupby("gender").pivot("dept",['IT','HR']).count().show()
```

## Stack()
The `stack()` function in PySpark is used to convert columns into rows, effectively unpivoting a DataFrame. This is useful
 when you want to transform wide data into a long format.

**Syntax:**
python
from pyspark.sql.functions import expr

df.selectExpr("id", "stack(n, col1, col2, ..., coln) as (new_col1, new_col2, ..., new_coln)")


**Example:**
Suppose you have a DataFrame with quarterly sales columns and want to unpivot them into rows:

python
data = [(1, 100, 200, 300, 400)]
columns = ["id", "Q1", "Q2", "Q3", "Q4"]
df = spark.createDataFrame(data, columns)
df.selectExpr("id", "stack(4, 'Q1', Q1, 'Q2', Q2, 'Q3', Q3, 'Q4', Q4) as (Quarter, Sales)").show()


This will convert the columns Q1, Q2, Q3, Q4 into rows with columns `Quarter` and `Sales`.

```python
from pyspark.sql.functions import expr
data = [('IT',2,3),('HR',4,5),(None,6,7)]
df1 = spark.createDataFrame(data,["dept","male","female"])
df2 = df1.select('dept',expr("stack(2,'Male',male,'Female',female) as (gender,count)")).show()
```
